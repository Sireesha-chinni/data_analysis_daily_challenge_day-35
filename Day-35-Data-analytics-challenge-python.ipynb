# Coding Challenge : Built in Function Python

Challenge 1: Flatten a Nested List
    
    
Concepts: list comprehension
    
    
Problem:
    
    
nested = [[1, 2], [3, 4], [5]]


Expected Output:
    
    
[1, 2, 3, 4, 5]

```python
nested = [[1, 2], [3, 4], [5]]
result = []

for i in nested:
    for j in i:
        result.append(j)
        
print(result)
```

Challenge 2: Check Anagram
    
    
Concepts: sorted()
    
Problem:
    
    
Check if two strings are anagrams.


s1 = "listen"


s2 = "silent"


Expected Output:
    
    
True

```python
s1 = "listen"

s2 = "silent"

print(sorted(s1) == sorted(s2))
```

Challenge 3: Find Missing Number
    
    
Concepts: set()
    
    
Problem:
    
    
nums = [1, 2, 4, 5]


Expected Output:
    
3

Hint:
    
    
Compare with full range using set difference.

```python
nums = [1, 2, 4, 5]

rst_set = set()

for i in range(1,6):
    rst_set.add(i)
    
nums = set(nums)

result = rst_set - nums

print(result)
```

Challenge 4: Group Words by First Letter
    
    
Concepts: dict, setdefault()
    
    
Problem:
    
    
words = ["apple", "ant", "banana", "bat"]


Expected Output:
    
    
{'a': ['apple', 'ant'], 'b': ['banana', 'bat']}

```python
words = ["apple", "ant", "banana", "bat"]

groups = {}

for word in words:
    groups.setdefault(word[0], []).append(word)

print(groups)
```

Challenge 5: Remove Empty Values
    
    
Concepts: dictionary comprehension
    
    
Problem:
    
    
data = {"a": 1, "b": None, "c": 3, "d": None}


Expected Output:
    
    
{'a': 1, 'c': 3}

```python
data = {"a": 1, "b": None, "c": 3, "d": None}

result = { k:v for k,v in data.items() if v is not None}


print(result)
```
